```python
import numpy as np
from knn import KNN
import matplotlib.pyplot as plt
```

```python
X_train = np.array([
    [1, 1],
    [1, 2],
    [2, 1],
    [8, 8],
    [8, 9],
    [9, 8]
])

y_train = np.array([3, 2, 0, 1, 2, 6])

X_test = np.array([7, 8])
```

```python
plt.scatter(X_train[:, 0], X_train[:, 1])
```

```python
plt.scatter(X_test[0], X_test[1])
```

```python
model = KNN(k = 3, distance_metric = "Manhattan")
model.fit(X_train, y_train)
model.predict(X_test, problem = "Regression")
```

```python
plt.scatter(X_train[:, 0], X_train[:, 1])
plt.scatter(X_test[0], X_test[1])

for point in X_train:
    plt.plot(
        [X_test[0], point[0]],
        [X_test[1], point[1]]
    )

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```
