```python
from naive_bayes import GaussianNaiveBayes 
import matplotlib.pyplot as plt
import numpy as np
```

```python
X = np.array([
    [20, 170],   # class 0
    [40, 180],   # class 1
    [30, 175],   # class 2
    [22, 172],   # class 0
    [42, 182],   # class 1
    [32, 177],   # class 2
    [24, 174],   # class 0
    [44, 184],   # class 1
    [36, 179]    # class 2
])

y = np.array([
    0,
    1,
    2,
    0,
    1,
    2,
    0,
    1,
    2
])
model = GaussianNaiveBayes()
```

```python
x_new = [37, 178]

model.fit(X, y)

prediction = model.predict(x_new)

print("Prediction:", prediction)
```
