```python
from logistic_regression import LogisticRegression
import matplotlib.pyplot as plt
import numpy as np
```

```python
X = np.array([1, 2, 3])
y = np.array([0, 0, 1])
```

```python
model = LogisticRegression(lr = 0.1, epochs = 100)
model.fit(X, y)
model.predict(X)
```

```python
plt.scatter(X, y)

plt.xlabel("X")
plt.ylabel("y")
plt.title("Training Data")

plt.show()
```

```python
x_values = np.linspace(0, 4, 100)

z = model.w * x_values + model.b
p = model.sigmoid(z)

plt.scatter(X, y)
plt.plot(x_values, p)

plt.axhline(0.5, linestyle="--")

plt.xlabel("X")
plt.ylabel("Probability")
plt.title("Logistic Regression")

plt.show()
```
